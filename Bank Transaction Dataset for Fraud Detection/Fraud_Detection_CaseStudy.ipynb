# 1. EDA (Exploratory data analysis) - “Bank Transaction Dataset for Fraud Detection”

The dataset used for this EDA (Exploratory data analysis) study is an open dataset, named “Bank Transaction Dataset for Fraud Detection”, which can be found in Kaggle.
The dataset contains 2512 rows, 16 columns of data of bank transaction record for customers, including the transaction amount, transaction type, etc. for analysis. The dataset provides great opportunities for analyzing customer transaction behavior, providing insight for utilizing bank resources, examining the potential security issue and fraud detection analysis.

Following the same license as the author Dataset: https://www.kaggle.com/datasets/valakhorasani/bank-transaction-dataset-for-fraud-detection/data


The dataset composed of 16 columns, details for each column are being described below:

*   TransactionID : Unique, alphanumeric - Identifier for each transaction
*   AccountID : alphanumeric - Identifier for each account number
*   TransactionAmount : numeric - Transaction amount for each transaction, this could be a great feature for analysis
*   TransactionDate : date - Date of transaction
*   TransactionType : text - Debit / Credit transaction
*   Location : text - Geographical location where the transaction occurs
*   DeviceID : alphanumeric - Identifier for the device used for transaction
*   IP Address : numeric - IP address for the transaction occurs
*   MerchantID : alphanumeric - Identifier for merchants
*   Channel : text - Transaction channel: Online / ATM / Branch
*   CustomerAge : numeric - Age of customer
*   CustomerOccupation : text - Occupation of customer
*   TransactionDuration : numeric - Duration of transaction used
*   LoginAttempts : numeric - Number of login attempts to complete the transaction, this could be a great feature for analysis
*   AccountBalance : numeric - Balance of customer account
*   PreviousTransactionDate : date - Last transaction date of the customer

# 2. The use of data mining algorithms for Bank Transaction Fraud Detection

In this study, mining algorithms for clustering will be selected to investigate how it performs on detecting fraud transaction from bank transaction record dataset. DBSCAN and OPTICS clustering algorithms are used here, DBSCAN and OPTICS algorithms are able to detect outliers. During the model training, the parameters and features will be adjusted to investigate how they would affect the result.

The result of the model provided would be the outliers in the dataset, which means they are different from others transaction that clustered by the algorithms. They could be know as potential fraud transaction.
For clusters that are not outliers, they could be the cluster of customers that have similar transaction behavior.

Other than fraud transaction detection, based on different objectives, it is possible to achieve other results such as customer transaction behavior study by using different algorithms with different selected features.

---

**Import library for loading dataset**

```python
import numpy as np
import pandas as pd
import os
```

```python
# loading dataset for data mining
df_bankcsv = pd.read_csv("dataset/bank_transactions_data_2.csv")
```

**Data featuring**

Convert the date entry to datetime format. Then add a column for showing how many days since last transaction made by the customer.

```python
# convert to datetime format for date value
df_bankcsv['TransactionDate'] = pd.to_datetime(df_bankcsv['TransactionDate'])
df_bankcsv['PreviousTransactionDate'] = pd.to_datetime(df_bankcsv['PreviousTransactionDate'])
# calculate the number of days since last transaction made by this customer
df_bankcsv['DaysSinceLastTransaction'] = (df_bankcsv['TransactionDate'] - df_bankcsv['PreviousTransactionDate']).dt.days
df_bankcsv.info()
```

---

**Fraud Transaction Detection**

DBSCAN clustering

4 of the features are selected, they are the ['AccountBalance', 'TransactionAmount', 'TransactionDuration', 'LoginAttempts'].

```python
from sklearn.cluster import DBSCAN
from sklearn.preprocessing import StandardScaler
from numpy import unique, where
from matplotlib import pyplot as plt

# Standardize the features
stdscaler = StandardScaler()
X_stdized = stdscaler.fit_transform(df_bankcsv[['AccountBalance', 'TransactionAmount', 'TransactionDuration', 'LoginAttempts']].values)

# DBSCAN parameters
dbscan_model = DBSCAN(eps=0.5, min_samples=5)
y_pred = dbscan_model.fit_predict(X_stdized)

# show number of outliers
dbscan_labels = dbscan_model.labels_
count_outliers = (dbscan_labels == -1).sum()
print(f"Number of outliers clustered by DBSCAN model: {count_outliers}")

# Retrieve the cluster formed
dbclusters = unique(y_pred)

# Create scatter plot for samples from each cluster
for dbcluster in dbclusters:
    data_pt = where(y_pred == dbcluster)
    if dbcluster == -1:
        plt.scatter(X_stdized[data_pt, 0], X_stdized[data_pt, 1], marker='x', label='Outliers', color='Black')
    else:
        plt.scatter(X_stdized[data_pt, 0], X_stdized[data_pt, 1], label=f'Cluster {dbcluster}', alpha=0.6)

# label for X & Y axis
xAxis = 'AccountBalance'
yAxis = 'TransactionAmount'

plt.title('DBSCAN Clustering')
plt.xlabel(xAxis)
plt.ylabel(yAxis)
plt.legend()
plt.show()
```

The result shows 2 cluster are formed by the model. There are 186 transactions that are outliers. It seems to be too many, around 7.4% of the transactions are classified as potential fraud.

The parameter is being adjusted to see if any change in output. In this case, the eps is adjusted from 0.5 to 1.5.

```python
from sklearn.cluster import DBSCAN

# Standardize the features
stdscaler = StandardScaler()
X_stdized = stdscaler.fit_transform(df_bankcsv[['AccountBalance', 'TransactionAmount', 'TransactionDuration', 'LoginAttempts']].values)

# DBSCAN parameters
dbscan_model = DBSCAN(eps=1.5, min_samples=5) # changing eps to 1.5
y_pred = dbscan_model.fit_predict(X_stdized)

# show number of outliers
dbscan_labels = dbscan_model.labels_
count_outliers = (dbscan_labels == -1).sum()
print(f"Number of outliers clustered by DBSCAN model: {count_outliers}")

# Retrieve the cluster formed
dbclusters = unique(y_pred)

# Create scatter plot for samples from each cluster
for dbcluster in dbclusters:
    data_pt = where(y_pred == dbcluster)
    if dbcluster == -1:
        plt.scatter(X_stdized[data_pt, 0], X_stdized[data_pt, 1], marker='x', label='Outliers', color='Black')
    else:
        plt.scatter(X_stdized[data_pt, 0], X_stdized[data_pt, 1], label=f'Cluster {dbcluster}', alpha=0.6)

plt.title('DBSCAN Clustering')
plt.xlabel(xAxis)
plt.ylabel(yAxis)
plt.legend()
plt.show()
```

The result shows 17 out of 2512 transactions are outliers. It seems better than the previous setting, around 0.68% of the transactions are classified as potential fraud.

The list below displays the potential fraud transactions, labeled by the column ['DBCluster'] with the value "-1".

From the data, it is obvious that transactions with more than 1 login attempt are more likely to be classified as outliers. (The total number of transactions where LoginAttempts > 1 is 122 in the dataset.)


```python
# create column in the dataframe for labelling
df_bankcsv['DBCluster'] = dbscan_model.labels_
# outliers is labelled as cluster "-1"
outliers_db = df_bankcsv[df_bankcsv['DBCluster'] == -1]
print(f"Potential frauds transaction detected by DBSCAN model: {len(outliers_db)}")
display(outliers_db)
```

OPTICS clustering

Same features are being selected for learning using OPTICS.

```python
from sklearn.cluster import OPTICS

# Standardize the features
stdscaler = StandardScaler()
X_stdized = stdscaler.fit_transform(df_bankcsv[['AccountBalance', 'TransactionAmount', 'TransactionDuration', 'LoginAttempts']].values)

# OPTICS parameters
optics_model = OPTICS(min_samples=5, xi=0.1, min_cluster_size=0.01)
y_pred = optics_model.fit_predict(X_stdized)

# show number of outliers
optics_labels = optics_model.labels_
count_outliers = (optics_labels == -1).sum()
print(f"Number of outliers clustered by OPTICS model: {count_outliers}")

# Retrieve the cluster formed
optics_clusters = unique(y_pred)

# Scatter plot for each cluster
for optics_cluster in optics_clusters:
    data_pt = where(y_pred == optics_cluster)
    if optics_cluster == -1:
        # Plot outliers with a different marker
        plt.scatter(X_stdized[data_pt, 0], X_stdized[data_pt, 1], marker='x', label='Outliers', color='Black')
    else:
        plt.scatter(X_stdized[data_pt, 0], X_stdized[data_pt, 1], label=f'Cluster {optics_cluster}', alpha=0.6)
plt.title('OPTICS Clustering')
plt.xlabel(xAxis)
plt.ylabel(yAxis)
plt.legend()
plt.show()
```

The results show 70 transactions are outliers. Now we will try to add 1 more feature 'DaysSinceLastTransaction' indicating the number of days since last transaction made by the customer.
Let's examine if the output would vary while keeping the parameters unchanged.

```python
from sklearn.cluster import OPTICS

# Standardize the features
stdscaler = StandardScaler()
X_stdized = stdscaler.fit_transform(df_bankcsv[['AccountBalance', 'TransactionAmount', 'LoginAttempts', 'DaysSinceLastTransaction']].values) # added feature 'DaysSinceLastTransaction'

# OPTICS parameters
optics_model = OPTICS(min_samples=5, xi=0.1, min_cluster_size=0.01)
y_pred = optics_model.fit_predict(X_stdized)

# show number of outliers
optics_labels = optics_model.labels_
count_outliers = (optics_labels == -1).sum()
print(f"Number of outliers clustered by OPTICS model: {count_outliers}")

# Retrieve the cluster formed
optics_clusters = unique(y_pred)

# Scatter plot for each cluster
for optics_cluster in optics_clusters:
    data_pt = where(y_pred == optics_cluster)
    if optics_cluster == -1:
        # Plot outliers with a different marker
        plt.scatter(X_stdized[data_pt, 0], X_stdized[data_pt, 1], marker='x', label='Outliers', color='Black')
    else:
        plt.scatter(X_stdized[data_pt, 0], X_stdized[data_pt, 1], label=f'Cluster {optics_cluster}', alpha=0.6)
plt.title('OPTICS Clustering')
plt.xlabel(xAxis)
plt.ylabel(yAxis)
plt.legend()
plt.show()
```

The results show 12 transactions are outliers, which is 0.48% of the total transaction records are being classified as potential fraud transaction.

Here below displayed the list of those potential fraud transactions, labelled by a column ['OPTCluster'] with value "-1".

Similar to the DBSCAN's, it is obvious that when login attempt is larger than 1, more likely it will be classified as outliers. (Total number of transactions that LoginAttempts >1 is 122 in dataset).

OPTICS model classified 2 of the transactions as outliers, while they are not outliers in DBSCAN model.

```python
# create column in the datafram for labelling
df_bankcsv['OPTCluster'] = optics_model.labels_
# outliers is labelled as cluster "-1"
outliers_opt = df_bankcsv[df_bankcsv['OPTCluster'] == -1]
print(f"Potential frauds transaction detected by OPTICS model: {len(outliers_opt)}")
display(outliers_opt)
```

# 3. Conclusion

Both DBSCAN and OPTICS algorithms behaves differently when choosing features for learning. There is no strict answer to say which one is better, they could be a compensation of each other.

To make use the result generated by the models, the outlier data points could be used as classify the transaction as a potential fraud transaction. The police force or bank staff could achieve their objective in a short period of time, rather that using traditional ways, for example simply searching for high login attempts to investigate (remember there is a hundred transactions that the LoginAttempts are >1).

**Possible ways to predict fraud transaction in future**

If a transaction is an actual fraud transaction after investigation, adding a column and label it as fraud transaction. Once there is a certain number of labelled transactions, machine learning algorithms for classification could be used to predict the transaction is a fraud transaction or not.

**Strategical approach of using the dataset**

Making use of the transaction dataset could be step by step as follows:
1.  Perform clustering by algorithms like DBSCAN, K-Means etc.
2.  Identify outliers detected by those algorithms, and investigate if it is actually a fraud transaction.
3.  Label the transaction as Normal / Fraud transaction after investigation.
4.  Use the labelled dataset to perform classification machine learning, such as using Logistic Regression or Random Forest, for prediction of fraud transaction for future transactions.
